## Machine Learning Workflow on Diabetes Data 
### Predict the onset of diabetes based on medical diagnostic measures 

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Step 1 - Load Data and Data Exploration 
### Load the data into a pandas dataframe

```python
diabetes = pd.read_csv('data/pi_diabetes.csv')
print(diabetes.columns)
```

```python
diabetes.shape
```

```python
diabetes.head()
```

### Create feature and target vectors
<h4 style="color:red;">The dataset has values that are either missing or incorrect (such as a blood pressure or BMI with a value of zero). Let's remove these "bad" values from the dataset.</h4>

#### *Remove missing or bad data*

```python
diabetes_model = diabetes[(diabetes.BloodPressure != 0) & (diabetes.BMI != 0) & (diabetes.Glucose != 0)]
print(diabetes_model.shape)
```

#### *Create feature matrix and target vector*
      #quick way to remove all but one column
        df.loc[:, df.columns != col]

```python
#df.loc[:, df.columns != col]
# Features and Targete
feature_names = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
X = diabetes_model[feature_names].values
y = diabetes_model.Outcome.values
```

```python
type(X)
X.shape
```

## Step 2 - Split Data into Training and Testing Sets
### Using Train/Test split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=0)
```

```python
X_train.shape
```

## Step 4 - Model Selection
#### K-Nearest Neighbors Fit (K=15)
Import the KNN Classifier and use the built-in score method to calculate some metrics

```python
# model selection process
from sklearn.neighbors import KNeighborsClassifier
# pick K-nearest neighbors
knn = KNeighborsClassifier(n_neighbors=15)
```

```python
# print information on the model
print(knn)
```

## Step 5 - Train the Model
#### Train the model
Here we fit(train) the KNN model to the training datasets.


```python
knn.fit(X_train, y_train)
```

## Step 6 - Model Predictions
#### Now that the model has been trained, let's send it the test dataset to see what it predicts.
We will store the results of its predictions into a vector we call y_pred:

```python
y_pred = knn.predict(X_test)
```

```python
print("Test set predictions:\n {}".format(y_pred))
print("Actual true values:\n {}".format(y_test))
```

## Step 7 - Evaluating the K-Nearest Neighbors Model
We can evaluate the model's performance in several ways. Two of the simplest ways are illustrated below.

First we will use the built-in scoring method. We do this by passing the test sets, both the test features (X_test) and the associated test
set labels (y test) also called categories or classes.

```python
# let's get the number of neighbors used and print the accuracy score
k=knn.n_neighbors
print("Model's Accuracy on Test Set for k = {0} is:{1:0.2f} \n".format(k, knn.score(X_test, y_test)))
```

   - <b>Misclassifications</b>
   
      Let's see how many misclassifcations occured. To do this, as the model has already been trained, let's use the prediction vector, y_pred computed above and the help of numpy.
      
      The calculating of the misclassifications is simple and is performed as follows:
      
      
              *misclassifications = (y_pred != y_test).sum()*
          
      which compares the predictions to the actual test set target values.<b> Do you understand what is occurring here?</b>

```python
# let's get the number of neighbors used and print the accuracy score
k=knn.n_neighbors
print('Misclassified samples for k = {0} are {1}\n'.format(k, (y_pred != y_test).sum()))
```

### Let'see how K=1 ( Nearest Neighbor) Performs

```python
# model selection process
from sklearn.neighbors import KNeighborsClassifier
# pick K-nearest neighbors
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)
```

```python
print("Model's Accuracy on the Test Set for k = {0} is:{1:0.2f} \n".format(k, knn.score(X_test, y_test)))
```

### Prediction on a single "unknown"

Let's pick an "unknown" from the test set and see how our K=1 model performs. We take this as meaning that we are making a
prediciton using this model on new data that the model has not seen but we know the classs labels. We test it on a patient with the
following features

```python
#index of unknown
index=45
#reshape and assigned
X_new = X_test[index].reshape(1,-1)
print('shape of X_new is {}|'.format(X_new.shape))
print('Predicted value is {}'.format(knn.predict(X_new)[0]))
print('Actual value is {}'.format(y_test[index]))
```
